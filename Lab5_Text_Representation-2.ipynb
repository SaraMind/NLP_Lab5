### Task 1: Cosine Similarity
Use the Cosine Similarity method to determine how similar the following sentences are.

'This is the first document.',  
'This document is the second document.',  
'And this is the third one.',  
'Is this the first document?'  


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

sentences = [
    "This is the first document.",
    "This document is the second document.",
    "And this is the third one.",
    "Is this the first document?"
]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(sentences)

similarity_matrix = cosine_similarity(tfidf_matrix)

df_similarity = pd.DataFrame(
    similarity_matrix,
    index=[f"Sentence {i+1}" for i in range(len(sentences))],
    columns=[f"Sentence {i+1}" for i in range(len(sentences))]
)

print(df_similarity.round(3))
````

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

sentences = [
    "data science is one of the most important fields of science",
    "this is one of the best data science courses",
    "data scientists analyze data"
]

tfidf = TfidfVectorizer()
result = tfidf.fit_transform(sentences)

df_tfidf = pd.DataFrame(
    result.toarray(),
    columns=tfidf.get_feature_names_out(),
    index=[f"Sentence {i+1}" for i in range(len(sentences))]
)

print(df_tfidf.round(3))
````

````python
import pandas as pd
import re
import gensim
from nltk.tokenize import word_tokenize
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

file_path = "simpsons_script_lines.csv"

df = pd.read_csv(file_path, encoding="utf-8", low_memory=False)

print(df.head())

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"[0-9]", "", text)
    text = re.sub(r"[)(,”“.’$-]", "", text)
    return text

df = df.dropna(subset=["spoken_words"])
df["cleaned_text"] = df["spoken_words"].apply(clean_text)

tokens = [
    word_tokenize(sentence)
    for sentence in df["cleaned_text"]
    if sentence.strip()
]

Skip_gram_model = gensim.models.Word2Vec(
    tokens,
    min_count=1,
    vector_size=100,
    window=5,
    sg=1
)

print("Skip-Gram Word2Vec model trained successfully!")
print("Vocabulary size:", len(Skip_gram_model.wv))
````

### Task 4

Use: wv.most_similar() method to :

1.	Find the words similar to “homer”.

2.	Find the words similar to “marge”.

3. Find the words similar to “bart”



````python
for word in ["homer", "marge", "bart"]:
    print(f"\nWords similar to '{word}':")
    try:
        print(Skip_gram_model.wv.most_similar(word, topn=10))
    except KeyError:
        print(f"'{word}' is not in the model vocabulary.")
````

### Task 5

Use the wv.doesnt_match() method to :

1.	Find which of 'jimbo', 'milhouse’, and 'kearney’ does not belong to the list.

3.	Find the odd one among "nelson", "bart", and "milhouse".

4.	Find the odd one among ‘homer', 'patty', and ‘selma'.

*Hint: You need to pass the strings as List*

````python
word_groups = [
    ["jimbo", "milhouse", "kearney"],
    ["nelson", "bart", "milhouse"],
    ["homer", "patty", "selma"]
]

for words in word_groups:
    print(f"\nWord list: {words}")
    try:
        odd_word = Skip_gram_model.wv.doesnt_match(words)
        print("Odd word out:", odd_word)
    except KeyError as e:
        print(f"Word not found in vocabulary: {e}")
````
